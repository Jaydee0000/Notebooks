# Replacement Search for Luke Shaw


#### My goal with this mini-project is to find a decent replacement for Luke Shaw, he is a Left-Back in a 4-3-3 system under Michael Carrick.


It is evident that Manchester United will not find a replacement for Luke Shaw as he is a cheaper option for the club so we need to define what are the important stipulations for making this a feasible replacement study.


Manchester United have historically made very questionable decisions in their scouting department therefore we need some ground rules as to who we can scout. The transfer policy of manchester united is very evident.

```python
import http.client
from google.colab import userdata
conn = http.client.HTTPSConnection("v3.football.api-sports.io")

headers = {
    'x-apisports-key': userdata.get('API-football')
}

conn.request("GET", "/transfers?team=33", headers=headers)

res = conn.getresponse()
data = res.read()

print(data.decode("utf-8"))
```

```python
import json
import pandas as pd

temp = json.loads(data.decode("utf-8"))
all_transfers_json = temp['response']
at_df = pd.json_normalize(all_transfers_json)


at_df = at_df.explode("transfers", ignore_index=True)

normalized = pd.json_normalize(at_df['transfers'])


df = pd.concat(
    [at_df.drop(columns=['transfers']), normalized],
    axis=1
)

df

```

```python
import pandas as pd
df["date"] = pd.to_datetime(df["date"], format="%Y-%m-%d")

recent_transfers = df[
    df["date"] > "2018-01-01"
].reset_index(drop=True)

recent_transfers
```

```python
import numpy as np

unique_players = pd.Series(pd.unique(recent_transfers['player.id']))

unique_players.describe()
```

```python
unique_players.to_csv("/content/unique_players.csv", header=["unique_player_id"])
```

```python
# @title
unique_player_list = pd.read_csv("unique_players.csv", usecols=["unique_player_id"])
unique_player_list.insert(1, "Date_of_Birth", pd.NaT)

conn = http.client.HTTPSConnection("v3.football.api-sports.io")

headers = {
    'x-apisports-key': userdata.get('API-football')
}

for index, row in unique_player_list.iterrows():
  player_id = unique_player_list.loc[index, 'unique_player_id']

  try:
    conn.request("GET", f"/players/profiles?player={player_id}", headers=headers)

    res = conn.getresponse()
    data = res.read()

    temp = json.loads(data.decode("utf-8"))

    normalized_player = pd.json_normalize(temp['response'])

    val = normalized_player.loc[0, 'player.birth.date']

    unique_player_list.loc[index, 'Date_of_Birth'] = val
  except Exception as e:
    print(e)

unique_player_list.to_csv("unique_players_with_dob.csv")
```

```python
import pandas as pd
info = pd.read_csv("/content/unique_players_with_dob.csv", usecols=["unique_player_id", "Date_of_Birth"])
df = pd.DataFrame(data=info, columns=["unique_player_id", "Date_of_Birth"])
df = df.dropna()
df.reset_index(inplace=True, drop=True)

df
```

```python
# The goal is to add the age column to the recent_transfers, then sort by date and age and see what the projections look like
recent_transfers = recent_transfers.merge(df, left_on="player.id", right_on="unique_player_id")
recent_transfers.columns
```

```python
recent_transfers.drop(axis=1, columns=["unique_player_id"], inplace=True)
recent_transfers
```

```python
recent_transfers.columns
recent_transfers = recent_transfers.convert_dtypes()

```

```python


players_brought_in = recent_transfers[
    (recent_transfers["teams.in.id"] == 33)

]

players_brought_in.reset_index(inplace=True, drop=True)
players_brought_in

duplicates = players_brought_in[
    players_brought_in.duplicated(subset=["player.id"], keep=False)
]

duplicates
```

```python
# make a hash table where we will store the keys of players that have 0 minutes for the united
# if the player's id appears in the hash table, remove the row
# for each row, get the year a player came in, and the id of the player.
# call the api and check if they have minutes in that year. if they do not try checking if they have minutes in the year after (only if it is not 2026).
# if they have minutes, continue, otherwise put their id into the hash table. Repeat

players_wo_minutes = set()
index_wo_minutes = set()
players_w_minutes = set()

conn = http.client.HTTPSConnection("v3.football.api-sports.io")

headers = {
    'x-apisports-key': userdata.get('API-football')
}

for index, row in players_brought_in.iterrows():

  playingtime = False

  player_id = players_brought_in.loc[index, 'player.id']
  player_date = players_brought_in.loc[index, 'date']
  player_year = player_date.year

  if(players_brought_in.loc[index, 'player.id'] in players_wo_minutes or
     players_brought_in.loc[index, 'player.id'] in players_w_minutes):
    continue

  try:
    conn.request("GET", f"/players?id={player_id}&season={player_year}", headers=headers)
    res = conn.getresponse()
    data = res.read()
    temp = json.loads(data.decode("utf-8"))
    temp = temp['response']
    temp = temp[0]
    if(not temp):
      players_wo_minutes.add(player_id)
      index_wo_minutes.add(index)
      print("not player information found")
      continue
    stats = pd.json_normalize(temp['statistics'])

    for index2, row2 in stats.iterrows():
      if(row2['team.id'] == 33):
        if(row2['games.minutes'] > 0):
          playingtime = True
          players_w_minutes.add(player_id)
          break
    if(playingtime):
      continue
    elif(player_year < 2026):
      conn.request("GET", f"/players?id={player_id}&season={player_year+1}", headers=headers)
      res = conn.getresponse()
      data = res.read()
      temp = json.loads(data.decode("utf-8"))
      temp = temp['response']
      if(not temp):
        players_wo_minutes.add(player_id)
        index_wo_minutes.add(index)
        print("not player information found")
        continue
      temp = temp[0]
      stats = pd.json_normalize(temp['statistics'])

      for index3, row3 in stats.iterrows():
        if(row3['team.id'] == 33):
          if(row3['games.minutes'] > 0):
            playingtime = True
            players_w_minutes.add(player_id)
            break

        else:
          continue




      if(not playingtime):
        players_wo_minutes.add(player_id)
        index_wo_minutes.add(index)



  except Exception as e:
    print(e)
    print(player_id)
    print(index)
    players_wo_minutes.add(player_id)
    index_wo_minutes.add(index)





```

```python
players_with_minutes = players_brought_in[
    players_brought_in['player.id'].isin(players_w_minutes)
]
players_with_minutes.reset_index(inplace=True, drop=True)
players_with_minutes = players_with_minutes.drop_duplicates(subset=["player.id"], keep="first")
```

```python
players_with_minutes.to_csv("players_with_minutes.csv")
```

```python
players_with_minutes["age_of_transfer"] = (
    players_with_minutes["date"].dt.year - pd.to_datetime(players_with_minutes["Date_of_Birth"]).dt.year
)
```

```python
grouped = players_with_minutes.groupby(
    pd.Grouper(key="date", freq="YE-MAY")
)["age_of_transfer"].mean()

df = grouped.reset_index(drop=False)
df.insert(2, "season", "na")
df
```

```python
df = df.drop([0])
```

```python
for index, row in df.iterrows():
  df.loc[index, "season"] = f"{df.loc[index, "date"].year - 1}/{df.loc[index, "date"].year}"
```
